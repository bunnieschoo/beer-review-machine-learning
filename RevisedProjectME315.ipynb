# 6. Feature Enginerring
Regression
* review_overall
Classification
* Create quality categories using percentiles

Bottom 25%  → Low
Middle 50%  → Medium
Top 25%     → High

Within the BeerAdvocate community, ratings are commonly interpreted approximately as:
| Overall Rating | Interpretation            |
| -------------: | ------------------------- |
|        4.5–5.0 | Outstanding / Exceptional |
|       4.0–4.49 | Excellent / Very Good     |
|       3.5–3.99 | Good                      |
|       3.0–3.49 | Average                   |
|          < 3.0 | Below Average / Poor      |


I was going to use Binary Classification using a threshold of 4.0 as it is simpler to understand and gives a clearer interpretation of Accuracy, Precision, Recall, and F1-score. However, taking a look at the plot distribution of overall beer ratings, most beers are already rated 4.0 or above. In that case, my target variable may end up very imbalanced. It may be better to choose the threshold based on the data distribution (using percentiles).

Percentile-Based Categories:

Bottom 25% --> Low quality
Middle 50% --> Medium quality
Top 25% --> High quality
This approach creates class based on the data distribution, which can help balance the dataset.

# Classification Target (Percentile-Based Quality)

````python
# ==========================================
# Beer Quality Categories
# ==========================================

q1 = sample_df["review_overall"].quantile(0.25)
q3 = sample_df["review_overall"].quantile(0.75)

def quality(score):

    if score <= q1:
        return "Low"

    elif score >= q3:
        return "High"

    else:
        return "Medium"

sample_df["quality"] = sample_df["review_overall"].apply(quality)

sample_df["quality"].value_counts()
````

````python
# Function to create RMSE 95% CI

import numpy as np
from sklearn.utils import resample
from sklearn.metrics import mean_squared_error

def rmse_ci(y_true, y_pred, n_bootstrap=1000):

    rmse_scores = []

    for i in range(n_bootstrap):

        y_boot, pred_boot = resample(
            y_true,
            y_pred,
            replace=True,
            random_state=i
        )

        rmse = np.sqrt(mean_squared_error(y_boot, pred_boot))
        rmse_scores.append(rmse)

    lower = np.percentile(rmse_scores, 2.5)
    upper = np.percentile(rmse_scores, 97.5)

    return lower, upper
````

````python
import numpy as np
import pandas as pd
from sklearn.utils import resample
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# --- Task 2: Classification Performance Comparison ---

def classification_accuracy_ci(model, X_test, y_test, n_bootstrap=1000):
    accuracies = []
    preds = model.predict(X_test)

    for i in range(n_bootstrap):
        y_boot, pred_boot = resample(y_test, preds, replace=True, random_state=i)
        accuracies.append(accuracy_score(y_boot, pred_boot))

    lower = np.percentile(accuracies, 2.5)
    upper = np.percentile(accuracies, 97.5)
    return lower, upper

# Calculate CIs for both models using classification test data
log_acc = accuracy_score(y_test_cls, pred)
log_lower, log_upper = classification_accuracy_ci(logistic, X_test_cls, y_test_cls)

rf_acc = accuracy_score(y_test_cls, rf_pred)
rf_lower, rf_upper = classification_accuracy_ci(rf_classifier, X_test_cls, y_test_cls)

# Create the comparison table
classification_comparison = pd.DataFrame({
    "Model": ["Logistic Regression", "Random Forest"],
    "Accuracy (95% CI)": [
        f"{log_acc:.3f} ({log_lower:.3f}, {log_upper:.3f})",
        f"{rf_acc:.3f} ({rf_lower:.3f}, {rf_upper:.3f})"
    ],
    "Precision": [
        precision_score(y_test_cls, pred, average='weighted'),
        precision_score(y_test_cls, rf_pred, average='weighted')
    ],
    "Recall": [
        recall_score(y_test_cls, pred, average='weighted'),
        recall_score(y_test_cls, rf_pred, average='weighted')
    ],
    "F1-Score": [
        f1_score(y_test_cls, pred, average='weighted'),
        f1_score(y_test_cls, rf_pred, average='weighted')
    ]
})

classification_comparison = classification_comparison.round(3)
display(classification_comparison)

# Export to LaTeX
print("\n--- LaTeX Table Output ---")
print(classification_comparison.to_latex(index=False, caption="Classification Performance Comparison (Task 2) with 95% CI", label="tab:class_comparison"))
````

````python
from sklearn.preprocessing import StandardScaler

cluster_features = [
    'review_aroma',
    'review_appearance',
    'review_palate',
    'review_taste',
    'beer_abv'
]

# Create feature matrix
X_cluster = sample_df[cluster_features].copy()

# Fill missing ABV values
X_cluster['beer_abv'] = X_cluster['beer_abv'].fillna(
    X_cluster['beer_abv'].median()
)

# Standardise
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_cluster)

print(X_scaled.shape)
````

````python
# Find the optimal number of clusters
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

silhouette_scores = []

k_values = range(2, 11)

for k in k_values:

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(X_scaled)

    score = silhouette_score(X_scaled, labels)

    silhouette_scores.append(score)

optimal_k = k_values[silhouette_scores.index(max(silhouette_scores))]

print("Optimal k =", optimal_k)
````
